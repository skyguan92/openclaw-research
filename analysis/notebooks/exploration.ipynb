# OpenClaw Research — 探索性分析

对比 OpenClaw / Nous Hermes / Claude Code 在记忆架构、Token 效率、任务成功率三个维度的表现。

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from compare import load_data, flatten_metrics

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

```python
# 加载数据 — 改为 use_demo=False 使用真实数据
records = load_data(use_demo=True)
df = flatten_metrics(records)
df.head(10)
```

## 1. 记忆架构分析

```python
mem = df[df['dimension'] == 'memory']
if not mem.empty:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    sns.barplot(data=mem, x='agent', y='recall_accuracy', ax=axes[0], palette='viridis')
    axes[0].set_title('召回准确率')
    axes[0].set_ylim(0, 1)
    sns.barplot(data=mem, x='agent', y='hallucination_count', ax=axes[1], palette='rocket')
    axes[1].set_title('幻觉次数')
    plt.tight_layout()
    plt.show()
else:
    print('无记忆测试数据')
```

## 2. Token 效率分析

```python
tok = df[df['dimension'] == 'token_efficiency']
if not tok.empty:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    sns.barplot(data=tok, x='agent', y='tokens_total', ax=axes[0], palette='crest')
    axes[0].set_title('总 Token 消耗')
    sns.barplot(data=tok, x='agent', y='tool_calls_count', ax=axes[1], palette='flare')
    axes[1].set_title('工具调用次数')
    plt.tight_layout()
    plt.show()
else:
    print('无 Token 效率数据')
```

## 3. 任务成功率分析

```python
suc = df[df['dimension'] == 'task_success']
if not suc.empty:
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    # 成功率
    pass_rates = suc.groupby('agent')['result'].apply(lambda x: (x == 'pass').mean())
    pass_rates.plot(kind='bar', ax=axes[0], color=['#4CAF50', '#FF9800', '#2196F3'])
    axes[0].set_title('通过率')
    axes[0].set_ylim(0, 1)
    axes[0].tick_params(axis='x', rotation=0)
    # 质量评分
    sns.barplot(data=suc, x='agent', y='quality_score', ax=axes[1], palette='mako')
    axes[1].set_title('质量评分 (1-5)')
    axes[1].set_ylim(0, 5)
    plt.tight_layout()
    plt.show()
else:
    print('无任务成功率数据')
```
